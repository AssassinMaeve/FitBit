```python
import pandas as pd
import os
import warnings

warnings.filterwarnings("ignore", message="Could not infer format")

# ===================== CONFIG =====================
ROOT_DIR = "DataSet"
OUTPUT_DIR = "cleaned_output"
MAX_ROWS = 1_000_000

TARGET_PERSON = "Fitbit_ba"   # 👈 ONLY this person will be processed

IMPORTANT_KEYWORDS = [
    "heart",
    "sleep",
    "oxygen",
    "spo2",
    "stress",
    "temperature",
    "biometric",
    "activity",
    "zone",
    "readiness",
    "menstrual",
    "snore"
]
# =================================================

person_path = os.path.join(ROOT_DIR, TARGET_PERSON)

if not os.path.isdir(person_path):
    raise FileNotFoundError(f"Folder not found: {person_path}")

print(f"\n📂 Processing person: {TARGET_PERSON}")

person_output_dir = os.path.join(OUTPUT_DIR, TARGET_PERSON)
os.makedirs(person_output_dir, exist_ok=True)


def clean_dataframe(df):
    df.columns = (
        df.columns.str.strip()
        .str.lower()
        .str.replace(" ", "_")
    )

    for col in df.columns:
        if "date" in col or "time" in col:
            df[col] = pd.to_datetime(df[col], errors="coerce")

            # Excel-safe timezone removal
            if isinstance(df[col].dtype, pd.DatetimeTZDtype):
                df[col] = df[col].dt.tz_localize(None)

    df.dropna(how="all", inplace=True)
    df.drop_duplicates(inplace=True)
    return df


def safe_name(name):
    for ch in ['\\', '/', '*', '?', ':', '[', ']']:
        name = name.replace(ch, '_')
    return name.strip().lower()


# ===================== PROCESS CATEGORY BY CATEGORY =====================
for root, dirs, files in os.walk(person_path):
    if root == person_path:
        continue

    category = os.path.basename(root)
    category_lower = category.lower()

    # Only physiological folders
    if not any(k in category_lower for k in IMPORTANT_KEYWORDS):
        dirs[:] = []
        continue

    print(f"   ➤ Processing category: {category}")

    dfs = []

    for file in files:
        if not file.lower().endswith((".csv", ".xlsx")):
            continue
        if "readme" in file.lower():
            continue

        file_path = os.path.join(root, file)

        try:
            df = pd.read_csv(file_path) if file.lower().endswith(".csv") else pd.read_excel(file_path)
        except Exception as e:
            print(f"     ❌ Failed to read {file}: {e}")
            continue

        df = clean_dataframe(df)
        if df.empty:
            continue

        df["source_file"] = file
        dfs.append(df)

    if not dfs:
        print(f"     ⚠️ No usable data in {category}")
        continue

    final_df = pd.concat(dfs, ignore_index=True)

    # Fill only numeric columns
    num_cols = final_df.select_dtypes(include="number").columns
    final_df[num_cols] = final_df[num_cols].fillna(0)

    output_file = os.path.join(
        person_output_dir,
        f"{TARGET_PERSON}_{safe_name(category)}.xlsx"
    )

    with pd.ExcelWriter(output_file, engine="openpyxl") as writer:
        total_rows = len(final_df)

        if total_rows <= MAX_ROWS:
            final_df.to_excel(writer, sheet_name="data", index=False)
        else:
            for i in range((total_rows // MAX_ROWS) + 1):
                start = i * MAX_ROWS
                end = start + MAX_ROWS
                chunk = final_df.iloc[start:end]
                chunk.to_excel(
                    writer,
                    sheet_name=f"data_{i+1}",
                    index=False
                )

    print(f"     ✅ Saved: {output_file}")
```

```python
import pandas as pd
import os
import warnings

warnings.filterwarnings("ignore", message="Could not infer format")

# ===================== CONFIG =====================
ROOT_DIR = "DataSet"
OUTPUT_DIR = "cleaned_output"
MAX_ROWS = 1_000_000

TARGET_PERSON = "Fitbit_vid"   # 👈 ONLY this person will be processed

IMPORTANT_KEYWORDS = [
    "heart",
    "sleep",
    "oxygen",
    "spo2",
    "stress",
    "temperature",
    "biometric",
    "activity",
    "zone",
    "readiness",
    "menstrual",
    "snore"
]
# =================================================

person_path = os.path.join(ROOT_DIR, TARGET_PERSON)

if not os.path.isdir(person_path):
    raise FileNotFoundError(f"Folder not found: {person_path}")

print(f"\n📂 Processing person: {TARGET_PERSON}")

person_output_dir = os.path.join(OUTPUT_DIR, TARGET_PERSON)
os.makedirs(person_output_dir, exist_ok=True)


def clean_dataframe(df):
    df.columns = (
        df.columns.str.strip()
        .str.lower()
        .str.replace(" ", "_")
    )

    for col in df.columns:
        if "date" in col or "time" in col:
            df[col] = pd.to_datetime(df[col], errors="coerce")

            # Excel-safe timezone removal
            if isinstance(df[col].dtype, pd.DatetimeTZDtype):
                df[col] = df[col].dt.tz_localize(None)

    df.dropna(how="all", inplace=True)
    df.drop_duplicates(inplace=True)
    return df


def safe_name(name):
    for ch in ['\\', '/', '*', '?', ':', '[', ']']:
        name = name.replace(ch, '_')
    return name.strip().lower()


# ===================== PROCESS CATEGORY BY CATEGORY =====================
for root, dirs, files in os.walk(person_path):
    if root == person_path:
        continue

    category = os.path.basename(root)
    category_lower = category.lower()

    # Only physiological folders
    if not any(k in category_lower for k in IMPORTANT_KEYWORDS):
        dirs[:] = []
        continue

    print(f"   ➤ Processing category: {category}")

    dfs = []

    for file in files:
        if not file.lower().endswith((".csv", ".xlsx")):
            continue
        if "readme" in file.lower():
            continue

        file_path = os.path.join(root, file)

        try:
            df = pd.read_csv(file_path) if file.lower().endswith(".csv") else pd.read_excel(file_path)
        except Exception as e:
            print(f"     ❌ Failed to read {file}: {e}")
            continue

        df = clean_dataframe(df)
        if df.empty:
            continue

        df["source_file"] = file
        dfs.append(df)

    if not dfs:
        print(f"     ⚠️ No usable data in {category}")
        continue

    final_df = pd.concat(dfs, ignore_index=True)

    # Fill only numeric columns
    num_cols = final_df.select_dtypes(include="number").columns
    final_df[num_cols] = final_df[num_cols].fillna(0)

    output_file = os.path.join(
        person_output_dir,
        f"{TARGET_PERSON}_{safe_name(category)}.xlsx"
    )

    with pd.ExcelWriter(output_file, engine="openpyxl") as writer:
        total_rows = len(final_df)

        if total_rows <= MAX_ROWS:
            final_df.to_excel(writer, sheet_name="data", index=False)
        else:
            for i in range((total_rows // MAX_ROWS) + 1):
                start = i * MAX_ROWS
                end = start + MAX_ROWS
                chunk = final_df.iloc[start:end]
                chunk.to_excel(
                    writer,
                    sheet_name=f"data_{i+1}",
                    index=False
                )

    print(f"     ✅ Saved: {output_file}")
```
